## loc and iloc indexing in a pandas DataFrame

The short summary is that `loc` indexing using labels, and `iloc` indexing uses integer position.  For rows these are often the same, but for columns they are typically different.

* Compare examples of `loc` and `iloc` using the penguins dataset.

* Aside: `loc` is one of the only Python examples I know where the right endpoint is included with slicing.

```python
import pandas as pd
```

```python
df = pd.read_csv("data/penguins.csv")
```

```python
df
```

```python
df[2,5]
```

```python
df.iloc[2,5]
```

```python
df.loc[2, "body_mass_g"]
```

```python
df
```

```python
# every row, columns 2-5, not including 5
df.iloc[:, 2:5]
```

```python
# right endpoint included with loc
df.loc[:, "bill_length_mm": "body_mass_g"]
```

```python
df.loc[:, ["bill_length_mm", "body_mass_g", "bill_length_mm", "bill_length_mm"]]
```

```python
# no column with the label 3
df[3]
```

```python
# with a slice, refers to row position (not to index label)
df[3:]
```

```python
df["body_mass_g"]
```

```python
df[["bill_length_mm", "body_mass_g", "bill_length_mm", "bill_length_mm"]]
```

## Boolean indexing

It was pointed out that the numeric values in the penguins dataset did not seem normally distributed.  What if we restrict to one particular species of penguin?  We will use a new concept, Boolean indexing, to answer.

* Using Seaborn's `kdeplot`, show the distribution of the bill lengths among all penguins in our penguins.csv dataset.

```python
import seaborn as sns
```

```python
sns.kdeplot(x = df["bill_length_mm"])
```

```python
sns.kdeplot(x = "bill_length_mm", data = df)
```

* Now show the same sort of plot, but restricting only to the "Adelie" species, using Boolean indexing.

```python
df
```

```python
df["species"]
```

```python
# boolean Series
df["species"] == "Adelie"
```

```python
# sub-DataFrame of Adelie penguins
df_sub = df[df["species"] == "Adelie"]
display(df_sub)
```

```python
# sub-DataFrame of Adelie penguins
df_sub = df.loc[df["species"] == "Adelie"]
display(df_sub)
```

```python
df["species"].value_counts()
```

```python
df_gentoo = df.loc[df["species"] == "Gentoo"]
display(df_gentoo)
```

```python
# note, this slicing corresponds to integer position
df_gentoo[3:]
```

* Instead of using Boolean indexing, use the `query` method.

```python
# getting Adelie penguins using `query`
df.query("species == 'Adelie'")
```

```python
# much closer to normally distributed with only one penguins species
sns.kdeplot(x = "bill_length_mm", data = df_sub)
```

* What is the average body mass among "Gentoo" penguins with a bill depth of less than 14?  Use Boolean indexing with a logical operator.

```python
# wrong `and`
ser = (df["species"] == "Gentoo") and (df["bill_depth_mm"] < 14)
```

```python
# use & for entry-by-entry comparison
ser = (df["species"] == "Gentoo") & (df["bill_depth_mm"] < 14)
ser
```

```python
ser.sum()
```

```python
df.loc[ser]
```

```python
df.loc[ser, "body_mass_g"]
```

```python
df.iloc[ser, 5]
```

```python
# Final answer
df.loc[ser, "body_mass_g"].mean()
```

* Same question but use `query`.

```python
df2 = df.query("(species == 'Gentoo') & (bill_depth_mm < 14)")
df2["body_mass_g"].mean()
```

```python
# comparing two columns, rather than a column to a constant
df.query("bill_length_mm < 1.8*bill_depth_mm")
```
