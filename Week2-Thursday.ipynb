````python
lengths = range(10**6, 10000000, 500000)

fig, axs = plt.subplots(2,1)
x = []
y = []
y2 = []

for lg in lengths:
    x.append(lg)

    mylist = list(range(lg))
    start = time.perf_counter()
    lg-6 in mylist
    elapsed = time.perf_counter() - start
    y.append(elapsed)

    # all values equal 4
    mydct = dict(zip(range(lg), np.full(lg, 4)))
    start = time.perf_counter()
    mydct[lg-6]
    elapsed = time.perf_counter() - start
    y2.append(elapsed)

axs[0].plot(x,y)
axs[0].set_title("search list")
axs[1].plot(x,y2)
axs[1].set_title("look up in dictionary")
````
